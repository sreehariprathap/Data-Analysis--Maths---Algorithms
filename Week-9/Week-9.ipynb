### Data Cleansing Process for User Data (Talent Acquisition) from a CSV File

```python
import pandas as pd
import matplotlib.pyplot as plt

file = "../data/Salary_Data.csv"
df_unfiltered = pd.read_csv(file)
row_count = len(df_unfiltered)

print("Total number of datapoints(rows)", row_count)
```

##### Removing empty valued rows from the dataset
By giving axis as 0 we filter rows with any empty values. 

```python
import numpy as np
import warnings
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
warnings.filterwarnings('ignore')

missing_values_before = df_unfiltered.isnull().sum()
print(missing_values_before)
print(df_unfiltered.describe())
print(missing_values_before.shape)

# Mean imputation
df_unfiltered['Age'].fillna(df_unfiltered['Age'].mean(), inplace=True)

# Median imputation
df_unfiltered['Years of Experience'].fillna(df_unfiltered['Years of Experience'].mean(), inplace=True)

# Mode imputation
df_unfiltered['Salary'].fillna(df_unfiltered['Salary'].mean(), inplace=True)

missing_values_after_numerical_imputation = df_unfiltered.isnull().sum()
print(missing_values_after_numerical_imputation)


df_unfiltered['Gender'].fillna(df_unfiltered['Gender'].mode()[0], inplace=True)
df_unfiltered['Education Level'].fillna(df_unfiltered['Education Level'].mode()[0], inplace=True)
df_unfiltered['Job Title'].fillna(df_unfiltered['Job Title'].mode()[0], inplace=True)

missing_values_after_category_removal = df_unfiltered.isnull().sum()
print(missing_values_after_category_removal)

npData = np.empty((6))
npData = np.vstack((npData, missing_values_before ))
npData = np.vstack((npData, missing_values_after_numerical_imputation))

df = pd.DataFrame(npData, columns=['Age', 'Gender', 'Education Level', 'Job Title', 'Years of Experience', 'Salary'])
df['Type'] = [ 'After Imputation of Categoric','Before Removal', 'After Imputation of Numeric']
display(df)


# Calculate correlation matrix
corr_matrix = df_unfiltered.corr()

# Plot using seaborn heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', vmin=-1, vmax=1)
plt.title("Correlation Matrix")
plt.show()
```

### Data Filling Report
 
We assessed missing values in the employee salary dataset, revealing:
 
- **Age**: 2
- **Gender**: 2
- **Education Level**: 3
- **Job Title**: 2

- **Salary**: 5
 
To clean the data, we dropped rows with any missing values initially, but later opted for imputation to retain more data. We filled missing values in numerical columns using the mean:  
- **Years of Experience** and **Salary** were imputed with their respective means.  
- **Age** was also filled with its mean.  
 
For categorical columns, we filled missing entries using the mode:  
- **Gender**, **Job Title**, and **Education Level** were imputed with the most frequent values.  
 
Post-cleaning, all missing values were resolved, resulting in a complete dataset ready for analysis.
